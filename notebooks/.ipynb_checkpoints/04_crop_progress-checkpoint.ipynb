```python
from pathlib import Path
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
# Project root
BASE_DIR = Path.cwd().parent

# Data files
CP_FILE = BASE_DIR / "data" / "nass" / "crop_progress.csv"


df = pd.read_csv(CP_FILE)

df_cp.head()
```

Crop Progress View

```python
df_cp = df.copy()

#Add Week Columns

df_cp["Week Number"] = df_cp["Period"].str.extract(r"#(\d+)").astype("Int64")

df_cp["Month"] = pd.to_datetime(df["Week Ending"]).dt.month

df_cp["Category"] = (
    df_cp["Data Item"]
    .str.extract(r"MEASURED IN PCT (.+)$")[0]
    .str.strip()
)

df_cp["Type"] = (
    df_cp["Data Item"]
    .str.split(" - ", n=1)
    .str[0].str.strip()
)
```

```python
def create_crop_view(filtered, type_name, category_name):

    view = filtered[
        (filtered["Type"] == type_name) &
        (filtered["Category"] == category_name)
    ].copy()

    view = view.sort_values("Week Number")

    return view
```

```python
#filter function

def get_crop_view(filtered, type_name, category_name):

    return (
        filtered[
            (filtered["Type"] == type_name) &
            (filtered["Category"] == category_name)
        ]
        .sort_values(["Year", "Week Number"])
        .copy()
    )
```

```python
current_year = filtered["Year"].max()

historical = filtered[
    filtered["Year"].between(
        current_year - 5,
        current_year - 1
    )
]

five_year_stats = (
    historical
    .groupby(["Type", "Category", "Week Number"])["Value"]
    .agg(
        **{
            "5Y Average": "mean",
            "5Y Min": "min",
            "5Y Max": "max"
        }
    )
    .reset_index()
)

filtered = filtered.merge(
    five_year_stats,
    on=["Type", "Category", "Week Number"],
    how="left"
)
```

```python


def plot_crop_progress_seasonality(
    filtered,
    type_name,
    category_name,
    current_year=None
):
    # Determine current year
    if current_year is None:
        current_year = filtered["Year"].max()

    # Filter selected crop and category
    crop = filtered[
        (filtered["Type"] == type_name) &
        (filtered["Category"] == category_name)
    ].copy()

    # Separate historical years from current year
    historical = crop[
        crop["Year"].between(
            current_year - 5,
            current_year - 1
        )
    ].copy()

    current = crop[
        crop["Year"] == current_year
    ].copy()

    # Calculate historical weekly minimum and maximum
    stats = (
        historical
        .groupby("Week Number")["Value"]
        .agg(
            Minimum="min",
            Maximum="max"
        )
        .reset_index()
        .sort_values("Week Number")
    )

    # Current year's weekly observations
    current = (
        current[["Week Number", "Value"]]
        .rename(columns={"Value": str(current_year)})
        .sort_values("Week Number")
    )

    # Create chart
    fig = go.Figure()

    # Historical maximum: upper boundary
    fig.add_trace(
        go.Scatter(
            x=stats["Week Number"],
            y=stats["Maximum"],
            mode="lines",
            line=dict(width=0),
            showlegend=False,
            hoverinfo="skip"
        )
    )

    # Historical minimum: lower boundary + shaded range
    fig.add_trace(
        go.Scatter(
            x=stats["Week Number"],
            y=stats["Minimum"],
            mode="lines",
            line=dict(width=0),
            fill="tonexty",
            fillcolor="rgba(100, 149, 237, 0.20)",
            name="Previous 5 Years Range",
            hovertemplate=(
                "Week %{x}<br>"
                "Historical Minimum: %{y:.1f}%"
                "<extra></extra>"
            )
        )
    )

    # Current year
    fig.add_trace(
        go.Scatter(
            x=current["Week Number"],
            y=current[str(current_year)],
            mode="lines+markers",
            name=str(current_year),
            line=dict(color="#d62728", width=3),
            marker=dict(size=5),
            hovertemplate=(
                "Week %{x}<br>"
                f"{current_year}: "
                "%{{y:.1f}}%<extra></extra>"
            )
        )
    )

    # USDA-style presentation
    fig.update_layout(
        title=f"{type_name} – {category_name} Progress",
        template="plotly_white",
        height=600,
        hovermode="x unified",
        xaxis_title="Week Number",
        yaxis_title="Crop Progress (%)",
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=0
        ),
        margin=dict(l=60, r=30, t=100, b=60)
    )

    fig.update_xaxes(
        dtick=2,
        showgrid=False,
        zeroline=False
    )

    fig.update_yaxes(
        ticksuffix="%",
        rangemode="tozero",
        showgrid=True,
        gridcolor="rgba(128, 128, 128, 0.2)"
    )

    return fig
```

```python
fig = plot_crop_progress_seasonality(
    filtered,
    type_name="CORN",
    category_name="DENTED",
    current_year = 2026
    
)

fig.show()
```

```python
# ============================================================
# 1. SINGLE CATEGORY CHART
# ============================================================

def plot_crop_progress_seasonality(
    filtered,
    type_name,
    category_name,
    current_year=None
):

    # --------------------------------------------------------
    # Determine current year automatically
    # --------------------------------------------------------

    if current_year is None:
        current_year = datetime.now().year

    previous_year = current_year - 1


    # --------------------------------------------------------
    # Filter selected crop and category
    # --------------------------------------------------------

    crop = filtered[
        (filtered["Type"] == type_name) &
        (filtered["Category"] == category_name)
    ].copy()

    if crop.empty:
        return None


    # --------------------------------------------------------
    # Historical years
    # Previous 5 years INCLUDING previous year
    # --------------------------------------------------------

    historical = crop[
        crop["Year"].between(
            current_year - 5,
            current_year - 1
        )
    ].copy()


    # --------------------------------------------------------
    # Current year
    # --------------------------------------------------------

    current = crop[
        crop["Year"] == current_year
    ].copy()


    # --------------------------------------------------------
    # Previous year
    # --------------------------------------------------------

    previous = crop[
        crop["Year"] == previous_year
    ].copy()


    # --------------------------------------------------------
    # Historical weekly minimum / maximum
    #
    # This uses the previous 5 years INCLUDING previous year.
    # --------------------------------------------------------

    stats = (
        historical
        .groupby("Week Number")["Value"]
        .agg(
            Minimum="min",
            Maximum="max"
        )
        .reset_index()
        .sort_values("Week Number")
    )


    # ========================================================
    # CREATE CHART
    # ========================================================

    fig = go.Figure()


    # ========================================================
    # HISTORICAL MAXIMUM
    # ========================================================

    fig.add_trace(
        go.Scatter(
            x=stats["Week Number"],
            y=stats["Maximum"],
            mode="lines",
            line=dict(width=0),
            showlegend=False,
            hoverinfo="skip"
        )
    )


    # ========================================================
    # HISTORICAL MINIMUM + SHADED RANGE
    # ========================================================

    fig.add_trace(
        go.Scatter(
            x=stats["Week Number"],
            y=stats["Minimum"],
            mode="lines",
            line=dict(width=0),
            fill="tonexty",
            fillcolor="rgba(100, 149, 237, 0.20)",
            name="Previous 5 Years Range",
            hovertemplate=(
                "Week %{x}<br>"
                "Historical Minimum: %{y:.1f}%"
                "<extra></extra>"
            )
        )
    )


    # ========================================================
    # PREVIOUS YEAR
    # ========================================================

    if not previous.empty:

        previous = (
            previous[
                ["Week Number", "Value"]
            ]
            .sort_values("Week Number")
        )

        fig.add_trace(
            go.Scatter(
                x=previous["Week Number"],
                y=previous["Value"],
                mode="lines+markers",
                name=str(int(previous_year)),
                line=dict(
                    width=2
                ),
                marker=dict(
                    size=4
                ),
                hovertemplate=(
                    "Week %{x}<br>"
                    f"{int(previous_year)}: "
                    "%{y:.1f}%"
                    "<extra></extra>"
                )
            )
        )


    # ========================================================
    # CURRENT YEAR
    # ========================================================

    if not current.empty:

        current = (
            current[
                ["Week Number", "Value"]
            ]
            .sort_values("Week Number")
        )

        fig.add_trace(
            go.Scatter(
                x=current["Week Number"],
                y=current["Value"],
                mode="lines+markers",
                name=str(int(current_year)),
                line=dict(
                    color="#d62728",
                    width=3
                ),
                marker=dict(
                    size=5
                ),
                hovertemplate=(
                    "Week %{x}<br>"
                    f"{int(current_year)}: "
                    "%{{y:.1f}}%"
                    "<extra></extra>"
                )
            )
        )


    # ========================================================
    # LAYOUT
    # ========================================================

    fig.update_layout(
        title=(
            f"{type_name} – "
            f"{category_name} Progress"
        ),

        template="plotly_white",

        height=500,

        hovermode="x unified",

        xaxis_title="Week Number",

        yaxis_title="Crop Progress (%)",

        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=0
        ),

        margin=dict(
            l=60,
            r=30,
            t=100,
            b=60
        )
    )


    # ========================================================
    # X AXIS
    # ========================================================

    fig.update_xaxes(
        dtick=2,
        showgrid=False,
        zeroline=False
    )


    # ========================================================
    # Y AXIS
    # ========================================================

    fig.update_yaxes(
        ticksuffix="%",
        rangemode="tozero",
        showgrid=True,
        gridcolor="rgba(128, 128, 128, 0.2)"
    )


    return fig


# ============================================================
# 2. PLOT ALL CATEGORIES FOR SELECTED TYPE
# ============================================================

def plot_all_crop_progress_categories(
    filtered,
    type_name,
    current_year=None
):

    # --------------------------------------------------------
    # Determine current year
    # --------------------------------------------------------

    if current_year is None:
        current_year = filtered["Year"].max()


    # --------------------------------------------------------
    # Get categories belonging to selected Type
    # --------------------------------------------------------

    categories = sorted(
        filtered.loc[
            filtered["Type"] == type_name,
            "Category"
        ]
        .dropna()
        .unique()
    )


    # --------------------------------------------------------
    # Plot every category
    # --------------------------------------------------------

    for category_name in categories:

        fig = plot_crop_progress_seasonality(
            filtered=filtered,
            type_name=type_name,
            category_name=category_name,
            current_year=current_year
        )

        if fig is not None:
            fig.show()


# ============================================================
# 3. TYPE DROPDOWN
# ============================================================

type_options = sorted(
    filtered["Type"]
    .dropna()
    .unique()
)


type_dropdown = widgets.Dropdown(
    options=type_options,
    description="Type:",
    value=type_options[0] if type_options else None,
    layout=widgets.Layout(
        width="300px"
    )
)


# ============================================================
# 4. UPDATE CHARTS WHEN TYPE CHANGES
# ============================================================

def on_type_change(change):

    if change["name"] == "value":

        clear_output(
            wait=True
        )

        display(
            type_dropdown
        )

        plot_all_crop_progress_categories(
            filtered=filtered,
            type_name=change["new"]
        )


# ============================================================
# 5. CONNECT DROPDOWN
# ============================================================

type_dropdown.observe(
    on_type_change,
    names="value"
)


# ============================================================
# 6. INITIAL DISPLAY
# ============================================================

display(
    type_dropdown
)

plot_all_crop_progress_categories(
    filtered=filtered,
    type_name=type_dropdown.value
)
```
