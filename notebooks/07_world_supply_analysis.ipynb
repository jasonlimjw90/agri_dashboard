```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/wasde_processed_all_historicals.csv")
df["crop"] = pd.Categorical(
    df["crop"],
    categories=["oc", "nc"],
    ordered=True
)
```

US MoM Change on demand (filter by country)

```python
display_items = df[
    (df["commodity"] == "corn") &
    (df["snd_item"].isin(["domestic total", "exports", "ending stocks"])) &
    (df["table"] == "ww") &
    (df["country"].isin(["united states"]))]
```

```python
display_items.pivot_table(
    index = "Report Date",
    columns = ["country", "snd_item", "crop"],
    values = "Value"
)
```

Exports YoY change (filter by months)

```python
display_items_1 = df[
    (df["commodity"] == "corn") &
    (df["table"] == "ww") &
    (df["Month"] == 8) &
    (df["snd_item"].isin(["exports", "domestic total", "ending stocks"]))]
```

```python
table = display_items_1.pivot_table(
    index= "country",
    columns= ["snd_item", "crop year", "crop"],
    values= "Value"
)

table.sort_values(
    by = ("exports", 2026, "nc"),
    ascending = False
)
```
