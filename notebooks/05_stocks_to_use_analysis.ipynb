```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/oce-wasde-report-data-2026-09.csv")
```

```python
display_items = df[
    (df["Commodity"] == "Corn") &
    (df["Attribute"].isin([
        "Ending Stocks",
        "Domestic Total",
        "Exports"]))&
    (df["ReportTitle"] == "World Corn Supply and Use")
]

df["Commodity"].unique()
```

```python
display_items.pivot_table(    
    index="Region",
    columns=["Attribute", "MarketYear"],
    values="Value",
    aggfunc="first")
```

```python
display_items_1 = df[
    (df["Commodity"] == "Corn") &
    (df["Region"].isin(["World","United States", "China"])) &
    (df["Attribute"].isin(["Ending Stocks","Domestic Total",   "Exports"]))&
    (df["ReportTitle"] == "World Corn Supply and Use")&
    (df["ProjEstFlag"] == "Proj.")
    ]
```

```python
display_items_1.pivot(
    index="ReleaseDate",
    columns=["Attribute","Region"],
    values = "Value",
)
```
